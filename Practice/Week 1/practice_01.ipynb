### Demo 4: Functions

```python
def is_even(a):   
    if a % 2 == 0:
        return True
    else:
        return False
print(is_even(2))
print(is_even(3))
print(is_even(-2))
print(is_even(-3))
```

the `lambda` function

```python
is_even = lambda a: a % 2 == 0
print(is_even(2))
print(is_even(3))
print(is_even(-2))
print(is_even(-3))
```

### Numpy, a very brief intro

```python
# Uncomment the following line to install numpy if you haven't already
# !pip install numpy

import numpy as np
```

```python
#vectors and matrices stored in numpy arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
A = np.array([[1, 2, 3], [4, 5, 6]])

# elementwise operations
print(a+b)
print(a-b)
print(a/b)
print(a**2)

# what is it?
# Spoiler: not the matrix product
print(a*a)
print(A*A)
```

```python
# The real matrix product is done with the @ operator
print(a@a)  
print(A@a)
print(A@A.T)  #A.T is the transpose of A
```

```python

#indexing as before
print(a[0])
print(A[0, 0])

#also slicing as before
print(a[0:2])
print(A[0:2, 0:2])

#particularly useful for matrices: slicing with a single index
print(A[0])  #first row
print(A[:, 0])  #first column
```

```python
#universal functions
print(np.sqrt(2))
print(np.sin(np.pi/2))
print(np.log(10))
```

### Task 1
Write a function that finds the minimum in a list!

```python
def find_min(l):
    min = l[0]
    for i in range(len(l)):
        if l[i] < min:
            min = l[i]
    return min
print(find_min([-7,-6,4,5,-6,4,-8,6,-5]))
```

### Task
[Harry Potter books](https://github.com/amephraim/nlp)

Your task will be to analyize the words in Harry Potter and The Sorcerer's Stone.

Download the Harry Potter books from the provided link. Then read in the text of the first book.
1. Split the text to words. Take care of special characters like ,?!" etc. and capital letters.
2. How many different words are in the book? Use set!
3. What are the ten most common words? Use dictionary to story the word counts.

```python
# Got help mostly from StackExchange discussions and friends.
with open("C:\\VSCode\\Introduction to Numerical Algorithms\\Practice\\Week 1\\J. K. Rowling - Harry Potter 1 - Sorcerer's Stone.txt") as file:
    str = file.read().lower()
spt = ""
for i in str:
    if i.isalnum():
        spt += i
    else: 
        spt += " "
total = spt.split()
unique = set(total)
print(f"The total number of words in the text is {len(total)}.")
print(f"The total number of unique words in the text is {len(unique)}.")

# Help was needed from one of my classmates.
top = {}
for word in total:
    top[word] = top.get(word, 0) + 1
sorting = sorted(top.items(), key = lambda item: item[1], reverse = True)[:10]
for word, count in sorting:
    print(f"The word '{word}' was mentioned {count} times.")
```
