## The Gauss–Seidel method

An iterative way to solve linear equations, $A\mathbf{x}=\mathbf{b}$.
We start with a guess for $\mathbf{x}$ and improve it by updating one component at a time.

For example:

$$
\begin{aligned}
4x_1+x_2+2x_3 &= 4,\\
3x_1+5x_2+x_3 &= 7,\\
x_1+x_2+3x_3 &= 3.
\end{aligned}
$$

Express one unknown from each equation:

$$
x_1=\frac{4-x_2-2x_3}{4},\qquad
x_2=\frac{7-3x_1-x_3}{5},\qquad
x_3=\frac{3-x_1-x_2}{3}.
$$

Starting from $(0,0,0)$, the first iteration is:

1. $x_1=(4-0-0)/4=1$.
2. $x_2=(7-3\cdot1-0)/5=0.8$, using the new $x_1$.
3. $x_3=(3-1-0.8)/3=0.4$, using the new $x_1$ and $x_2$.

Now repeat the same steps starting from $(1,0.8,0.4)$.

**Always use the latest available values.** Updating the same array immediately gives this behavior. Using only values from the previous iteration would give the Jacobi method.

For a general row:

$$
x_i^{(k+1)}=\frac{1}{a_{ii}}\left(b_i-\sum_{j<i}a_{ij}x_j^{(k+1)}-\sum_{j>i}a_{ij}x_j^{(k)}\right).
$$

Convergence is guaranteed if $A$ is **strictly diagonally dominant by rows**, or **symmetric positive definite**. Strict diagonal dominance means $|a_{ii}|>\sum_{j\ne i}|a_{ij}|$ in every row. The example above satisfies this condition. Symmetry alone is not enough.
