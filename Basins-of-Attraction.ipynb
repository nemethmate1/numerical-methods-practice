## Part 2: Complex functions (2D)

Let us have a function $f(z)$, where $z$ is a complex number (represented as `j` in `python`). For the sake of understanding the problem first, let $f(z)$ be the one in the example:

$$ f(z) = z^3 - 1 = 0 \qquad \Longrightarrow \qquad z_0 = 1, \quad z_1 = \dfrac{-1 + \sqrt{3}\,i}{2}, \quad z_2 = \dfrac{-1 - \sqrt{3}\,i}{2}$$

````python
# The roots of the example are:
z_0 = 1
z_1 = (-1 + np.sqrt(3)*1j) / 2
z_2 = (-1 - np.sqrt(3)*1j) / 2

# Put the roots into a numpy array:
roots = np.array([z_0, z_1, z_2])
````

As a safety measure, we can check whether these roots are valid solutions by plugging them back into $f(z)$ (so checking for $|f(r)|$). Doing so will not give us exactly $0$, but a value very close to that. We can define a threshold to check whether the substitution of the respective root is *close enough* to $0$ or not. Let that be a threshold of $10^{-10}$ (well above `python`'s rounding errors, keeping the actual errors distinctive).

````python
for r in roots:
    print(r, abs(r**3 - 1))

tol = 1e-10
print(all(abs(r**3 - 1) < tol for r in roots))
````

Notice that in Newton's formula

$$ z_{n+1} = z_n - \frac{f(z_n)}{f^{\prime}(z_n)} $$

the denominator contains the function's derivative. We are explicitly asked to use ```sympy``` to perform symbolic derivation. It gives us an exact symbolic expression that we can convert to numerical values using `lambdify`, so that they behave as ordinary `python` functions.

````python
# Define the symbol:
z_symb = sp.Symbol('z')

# Define the function:
f = z_symb**3 - 1

# Find the derivative symbolically:
f_prime = sp.diff(f, z_symb)

print(f"Formula: {f_prime}")

# Convert them using lambdify:
f_num  = sp.lambdify(z_symb, f, 'numpy')
df_num = sp.lambdify(z_symb, f_prime, 'numpy')

# Compare our results as a sanity check using central differences:
z0 = 0.7 + 0.3j
h = 1e-6
print(df_num(z0))
print((f_num(z0 + h) - f_num(z0 - h)) / (2*h))
````

Using Newton's formula mentioned above, we need to construct a function that takes a starting point $z_0$ and runs the iteration.
- The **inputs** are `z0`, `MAX_ITERATION` (say 50) and `TOLERANCE` (say $10^{-10}$). The iteration updates until the maximum number of iterations is reached (ideally).
- The iteration can stop early if when the step is tiny, i.e. $|z_{n+1} - z_n| <$ `tol`, or when $|f(z_n)|<$ `tol`.
- We return the final value of $z$. It’s also useful to return the number of iterations used, because we can color by that later for nicer pictures.
- If $f'(z_n)$ is $0$ (or extremely close), we divide by zero (error). For $z^3 - 1$ this happens at $z = 0$.
