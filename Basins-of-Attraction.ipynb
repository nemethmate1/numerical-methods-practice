## Part 2: Complex functions (2D)

Let us have a function $f(z)$, where $z$ is a complex number (represented as `j` in `python`). For the sake of understanding the problem first, let $f(z)$ be the one in the example:

$$ f(z) = z^3 - 1 = 0 \qquad \Longrightarrow \qquad z_0 = 1, \quad z_1 = \dfrac{-1 + \sqrt{3}\,i}{2}, \quad z_2 = \dfrac{-1 - \sqrt{3}\,i}{2}$$

````python
# The roots of the example are:
z_0 = 1
z_1 = (-1 + np.sqrt(3)*1j) / 2
z_2 = (-1 - np.sqrt(3)*1j) / 2

# Put the roots into a numpy array:
roots = np.array([z_0, z_1, z_2])
````

As a safety measure, we can check whether these roots are valid solutions by plugging them back into $f(z)$ (so checking for $|f(r)|$). Doing so will not give us exactly $0$, but a value very close to that. We can define a threshold to check whether the substitution of the respective root is *close enough* to $0$ or not. Let that be a threshold of $10^{-10}$ (well above `python`'s rounding errors, keeping the actual errors distinctive).

````python
for r in roots:
    print(r, abs(r**3 - 1))

tol = 1e-10
print(all(abs(r**3 - 1) < tol for r in roots))
````

Notice that in Newton's formula

$$ z_{n+1} = z_n - \frac{f(z_n)}{f^{\prime}(z_n)} $$

the denominator contains the function's derivative. We are explicitly asked to use ```sympy``` to perform symbolic derivation. It gives us an exact symbolic expression that we can convert to numerical values using `lambdify`, so that they behave as ordinary `python` functions.

````python
# Define the symbol:
z_symb = sp.Symbol('z')

# Define the function:
f = z_symb**3 - 1

# Find the derivative symbolically:
f_prime = sp.diff(f, z_symb)

print(f"Formula: {f_prime}")

# Convert them using lambdify:
f_num  = sp.lambdify(z_symb, f, 'numpy')
df_num = sp.lambdify(z_symb, f_prime, 'numpy')

# Compare our results as a sanity check using central differences:
z0_check = 0.7 + 0.3j
h_check = 1e-6
print(df_num(z0_check))
print((f_num(z0_check + h_check) - f_num(z0_check - h_check)) / (2*h_check))
````

Using Newton's formula mentioned above, we need to construct a function that takes a starting point $z_0$ and runs the iteration.
- The **inputs** are the function `f` and its derivative `fp`, the numerical numbers `z_start` (the starting number in Newton's formula, equivalent to `z0`), `MAX_ITERATION` (say 50) and `TOLERANCE` (say $10^{-10}$). The iteration updates until the maximum number of iterations is reached (ideally).
- If we were to pass `f = f_num` and `fp = df_num` as function inputs, then for our own function (later) we would have to pass the new pair explicitly, otherwise the function would keep using the $z^3 - 1$ pair. To avoid this, we drop the defaults and just pass `f` and `fp` when calling.
- The iteration can stop early when the step is tiny, i.e. $|z_{n+1} - z_n| <$ `tol`, or when $|f(z_n)|<$ `tol`.
- We return the final value of $z$. It’s also useful to return the number of iterations used, because we can color by that later for nicer pictures.
- If $f'(z_n)$ is $0$ (or extremely close), we divide by zero (error). For $z^3 - 1$ this happens at $z = 0$. We can raise an error for these cases, avoiding such divisions.

````python
# Defining the function and setting default values for the constants:
def newton_method(z_start, f, fp, MAX_ITERATION = 50, TOLERANCE = 1e-10):
    z = z_start  # z is the current point; it starts at z_start
    for n in range(MAX_ITERATION):
        
        # Evaluating f and f' at the current z:
        fz = f(z)
        dfz = fp(z)
        
        # We need to check whether the derivative is close to zero and raise an error in case:
        if abs(dfz) < 1e-14:
            raise ValueError("Warning: derivative is (almost) zero!")
        
        # Define the new variable and state Newton's formula:
        z_new = z - (fz / dfz)
        
        # Check whether this step was tiny, meaning the iteration stops:
        if abs(z_new - z) < TOLERANCE:
            return z_new, (n + 1)  # The final z (z_new is the freshest), and how many iterations were used
        
        z = z_new  # We need to copy our value back so that the loop can succeed
    return z, MAX_ITERATION  # In case there are no early breaks (returns)

# We test the iteration at various points:
print(newton_method(2, f_num, df_num))
print(newton_method(-1 + 1j, f_num, df_num))
print(newton_method(1, f_num, df_num))  # start exactly at a root

# Explicitly check for a case thats impossible, and we expect an error:
try:
    newton_method(0, f_num, df_num)
except ValueError as e:
    print(e)
````

After the iteration stops, we need to decide **which root** the run reached. The final value $z_\text{final}$ is very close to a root, but not exactly equal to it, so we compare distances:

$$ d_k = \left| z_\text{final} - r_k \right|, \qquad k = 0, 1, 2 $$

where $r_k$ are the roots stored in `roots`.

Complex numbers are points in the plane (the real part is the $x$-coordinate and the imaginary part is the $y$-coordinate), and the size of their difference is the ordinary distance between the two points. For a complex number $a + bi$, the absolute value is $|a + bi| = \sqrt{a^2 + b^2}$, so $d_k$ tells us how far $z_\text{final}$ is from the root $r_k$.

In the code, `abs(roots - z_final)` computes all of these distances at once and stores them in an array called `distances`, with one entry per root, in the same order as `roots`.

We pick the index $k$ with the smallest distance (using `np.argmin`), and accept it only if $d_k <$ `TOLERANCE`. If no root is close enough, the function returns `-1`, meaning *did not converge to any root*.

````python
def which_root(z_final, roots, TOLERANCE = 1e-6):
    distances = abs(roots - z_final)  # one distance per root, all at once
    k = np.argmin(distances)  # position of the smallest distance
    if distances[k] < TOLERANCE:
        return k
    return -1

# We test it the same way we did for Newton's method:
z_final, n_iter = newton_method(2, f_num, df_num)
print(which_root(z_final, roots))  # expecting 0
z_final, n_iter = newton_method(-1 + 1j, f_num, df_num)
print(which_root(z_final, roots))  # expecting 1 or 2
print(which_root(5 + 5j, roots))  # expecting -1
````

To color the complex plane, we need a **grid of starting points**. Every point of the grid is a complex number $z_0 = x + iy$, and each one will become one pixel of the final picture.

Simply put, the complex plane is infinite, but a picture has to be finite, so we pick a piece of it to look at, where $x$ is the real part and $y$ is the imaginary part.

We choose a square window of the plane, $x, y \in [-2, 2]$, which comfortably contains all three roots (they lie on the unit circle, $|r_k| = 1$), and sample it with $N$ equally spaced points in each direction. This gives $N \times N$ starting points. We start with a small resolution ($N = 200$) so the computation stays fast, and increase it later for the high-resolution images.

The notation $x, y \in [-2, 2]$ means both go from $-2$ to $2$. That is a square with corners at $-2-2i$, $-2+2i$, $2-2i$ and $2+2i$, centered on the origin.

`np.linspace` creates the lists of real parts $x$ and imaginary parts $y$, and `np.meshgrid` turns them into two $N \times N$ arrays `X` and `Y`, so that `X[i, j]` and `Y[i, j]` are the real and imaginary parts of the point in row $i$ and column $j$. We combine them into one complex array:

$$ Z_{ij} = X_{ij} + i\,Y_{ij} $$

In other words, we cannot test every point of the complex plane, so we choose a square region, $-2 \le x \le 2$ and $-2 \le y \le 2$, and place a grid of $N \times N$ evenly spaced starting points in it. The region contains all three roots, since they are at distance $1$ from the origin. The number $N$ is the resolution of the picture: a larger $N$ gives a sharper image but takes longer to compute. We start with $N = 200$ to keep the computation fast, and increase it later.

Note that the rows of `Z` correspond to the imaginary part $y$ and the columns to the real part $x$. This matters when we plot the result as an image.

````python
N = 2000  # resolution: starting small, increasing later
x = np.linspace(-2, 2, N)
y = np.linspace(-2, 2, N)
X, Y = np.meshgrid(x, y)
Z = X + 1j*Y  # Z[i, j] is the starting point in row i, column j

# Checking
print(Z.shape)  # expecting (200, 200), as 'shape' returns (rows, columns) in this case
print(Z[0, 0])  # expecting the corner (-2-2j)
print(Z[-1, -1])  # expecting the opposite corner (2+2j)
````

Now we run Newton's method from **every** point of the grid. For each starting point $Z_{ij}$ we:

1. iterate until the step is tiny (or the iteration cap is reached),
2. use `which_root` to find which of the three roots the run reached,
3. store the root's index in the array `basins`, and the number of iterations in `iterations`.

If the derivative vanishes along the way, `newton_method` raises an error. We catch it, so one bad starting point does not stop the whole computation, and leave the entry at $-1$ ("did not converge"). The array `basins` is then displayed as an image, with one color per root and black for $-1$.

````python
# We need to store one number per gridpoint using an array:
basins = np.full(Z.shape, -1)  # -1 represents "no root"

# Storing the iteration counts in an array:
iterations = np.zeros(Z.shape, dtype=int)

# Looping over every row 'i' and column 'j':
for i in range(N):
    for j in range(N):
        try:
            # Unpacking and storing the values straight into the arrays (though it can make checking a bit harder)
            z_final, iterations[i, j] = newton_method(Z[i, j], f_num, df_num)  # starts at the grid point
            basins[i, j] = which_root(z_final, roots)  # finding which root it reached
        except ValueError:
            pass  # leaving basins[i, j] = -1 and iterations[i, j] = 0
        
# Checking the loop first:
print(basins.shape, "and", iterations.shape)  # both (200, 200)
print(np.unique(basins))  # expect [0 1 2], maybe also -1
print(iterations.min(), "and", iterations.max())

# Plotting our image:
from matplotlib.colors import ListedColormap
cmap = ListedColormap(['black', 'tab:red', 'tab:green', 'tab:blue'])
plt.imshow(basins + 1, cmap=cmap, vmin=0, vmax=3, extent=[-2, 2, -2, 2], origin='lower')
plt.xlabel(r'$\mathbf{\mathrm{Re}(z)}$')
plt.ylabel(r'$\mathbf{\mathrm{Im}(z)}$')
plt.title('Basins of attraction of Newton\'s method for $z^3 - 1$')
plt.show()
````

### An optional, but faster alternative: vectorization and high-quality rendering

The double loop is easy to read, but it is slow: it calls `newton_method` once per pixel, and a Python loop over millions of pixels takes minutes. Since `f_num` and `df_num` already accept whole `numpy` arrays, we can run **one Newton step on all starting points at once**, and only loop over the iterations (at most `MAX_ITERATION`), not over the pixels.

**The idea.** We keep a list of the pixels that are still iterating. In every round we:

1. evaluate $f'(z)$ on the active pixels and remove those where $|f'(z)| <$ `DERIV_MIN`. This is the array version of raising a `ValueError`, since `numpy` division by zero gives `inf` or `nan` and a warning, not an error,
2. compute the Newton step $f(z)/f'(z)$ for all remaining pixels and update their points,
3. remove the pixels whose step is smaller than `TOLERANCE`, because they have converged and need no more work.

The rules are exactly the same as in `newton_method`, so both versions should give the same result. The function `which_root_grid` is the array version of `which_root`: it loops over the three roots (not over the pixels), and for every pixel it keeps the index of the closest root, or $-1$ if no root is closer than `TOLERANCE`.

**Validation.** The loop version stays as the readable reference. We run both on the same grid and check that the results agree. We also time them, since the speedup is itself a result about the efficiency of the methods.

**High-quality image.** Raising the resolution $N$ alone does not make the picture look better, because the screen shows far fewer pixels than the data contains. Three things matter for a high-quality image:

- **Pixel count.** We render directly at the target size (e.g. 4K or 8K) and save as a lossless PNG, so one data pixel is one image pixel.
- **Supersampling.** Each pixel is colored by a single starting point, so boundaries look jagged. We compute $s \times s$ points per pixel and average their colors, which smooths the edges.
- **Shading by iteration count.** Starting points near a basin boundary need many steps to converge. Darkening the color with the iteration count outlines the fractal boundaries. Points that did not converge are black.

An 8K image with supersampling contains hundreds of millions of sample points, so `render_basins` works in horizontal **strips** and fills the final image one strip at a time to limit memory use.

This part of the project was an optional experiment to see if a more efficient method was possible. External resources, such as *AI* and *StackExchange* was used to help in some parts (`render_basins` and time validation), but the rest of the code was simply similar to the previous methods and not much change were needed.

````python
import time
from matplotlib.colors import to_rgb


def newton_grid(Z, f, fp, MAX_ITERATION=50, TOLERANCE=1e-10, DERIV_MIN=1e-14):
    """Newton's method on every starting point of Z at once.

    Same rules as newton_method: a pixel stops when its step is below
    TOLERANCE, and is flagged as failed when |f'(z)| < DERIV_MIN.
    Returns the final points, the iteration counts and a mask of failed pixels.
    """
    z = Z.astype(complex).ravel()
    iters = np.zeros(z.size, dtype=int)
    failed = np.zeros(z.size, dtype=bool)
    idx = np.arange(z.size)

    for n in range(MAX_ITERATION):
        zi = z[idx]
        dfz = fp(zi)

        bad = np.abs(dfz) < DERIV_MIN
        failed[idx[bad]] = True
        idx, zi, dfz = idx[~bad], zi[~bad], dfz[~bad]

        step = f(zi) / dfz
        z[idx] = zi - step
        iters[idx] = n + 1

        idx = idx[np.abs(step) >= TOLERANCE]
        if idx.size == 0:
            break

    iters[failed] = 0
    return z.reshape(Z.shape), iters.reshape(Z.shape), failed.reshape(Z.shape)


def which_root_grid(Z_final, roots, TOLERANCE=1e-6):
    """Array version of which_root: index of the nearest root, or -1."""
    best_dist = np.full(Z_final.shape, np.inf)
    best_k = np.full(Z_final.shape, -1)
    for k, r in enumerate(roots):
        d = np.abs(Z_final - r)
        closer = d < best_dist
        best_dist[closer] = d[closer]
        best_k[closer] = k
    best_k[best_dist >= TOLERANCE] = -1
    return best_k


def render_basins(roots, f, fp, W=3840, H=2160, half_h=2.0, center=0j,
                  s=3, strip=60, MAX_ITERATION=100):
    """Render the basins as an RGB image of W x H pixels.

    Each pixel averages s x s sample points (supersampling) to smooth the
    edges, and its brightness decreases with the iteration count, which
    outlines the fractal boundaries. Works in strips to limit memory use.
    """
    half_w = half_h * W / H
    dx, dy = 2 * half_w / (W * s), 2 * half_h / (H * s)
    palette = np.array([to_rgb(c) for c in
                        ['tab:red', 'tab:green', 'tab:blue',
                         'tab:orange', 'tab:purple', 'tab:cyan']])
    img = np.zeros((H, W, 3), dtype=np.uint8)
    xs = center.real - half_w + (np.arange(W * s) + 0.5) * dx

    for r0 in range(0, H, strip):
        r1 = min(r0 + strip, H)
        rows = np.arange(r0 * s, r1 * s)
        ys = center.imag + half_h - (rows + 0.5) * dy
        Z_strip = xs[None, :] + 1j * ys[:, None]

        Zf, its, failed = newton_grid(Z_strip, f, fp, MAX_ITERATION=MAX_ITERATION)
        k = which_root_grid(Zf, roots)
        k[failed] = -1

        shade = 0.35 + 0.65 * np.exp(-0.08 * np.maximum(its - 3, 0))
        rgb = palette[np.maximum(k, 0)] * shade[..., None]
        rgb[k < 0] = 0

        rgb = rgb.reshape(r1 - r0, s, W, s, 3).mean(axis=(1, 3))
        img[r0:r1] = (255 * rgb).astype(np.uint8)
    return img


# Validation against the loop version (uses Z, basins and iterations from above)
t0 = time.perf_counter()
Z_final_vec, iterations_vec, failed_vec = newton_grid(Z, f_num, df_num)
basins_vec = which_root_grid(Z_final_vec, roots)
basins_vec[failed_vec] = -1
t_vec = time.perf_counter() - t0

print("Same basins:", np.array_equal(basins_vec, basins))
print("Same iterations:", np.array_equal(iterations_vec, iterations))
print("Differing pixels:", np.sum(basins_vec != basins))
print(f"Vectorized time: {t_vec:.2f} s")

# High-quality images (the final render takes several minutes)
img = render_basins(roots, f_num, df_num, W=7680, H=4320, s=3)
plt.imsave('basins_z3_8k.png', img)
````
